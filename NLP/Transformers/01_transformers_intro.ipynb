## 1. Traditional NLP vs. Modern NLP: The Context Problem

In traditional NLP, word representations were **static**. For example, Word2Vec assigns a single fixed numerical vector to every word.

Consider these two sentences:
1. *"The **bank** approved my loan."* (Financial institution)
2. *"I sat near the **bank** of the river."* (Sloping riverbank)

In traditional static representations, the word **bank** gets the exact same vector in both sentences!

Modern NLP uses **Transformers** to build **contextual representations**, where the vector representation of *"bank"* changes dynamically based on surrounding words.



### 🔍 Concept Visualization

```text
Traditional NLP:  [bank] ───> Fixed Vector (Financial + River mixed)

Modern NLP:       "The bank approved my loan"  ───> [bank] ───> Financial Vector
                  "I sat near the bank..."     ───> [bank] ───> Riverbank Vector
```



````python
# Simple demonstration comparing word usage contextually
sentence1 = "The bank approved my loan."
sentence2 = "I sat near the bank of the river."

print("Sentence 1:", sentence1)
print("Sentence 2:", sentence2)
print("\nIn sentence 1, 'bank' relates to financial context (loan, approved).")
print("In sentence 2, 'bank' relates to nature/geography context (river, sat).")

````

## 2. Intuition Behind Key Transformer Concepts

### 1. Context & Self-Attention
Self-attention allows a model to look at *all words in a sentence simultaneously* and pay more "attention" to words that clarify the meaning of other words.

- In *"The bank approved my loan"*, **"bank"** pays strong attention to **"approved"** and **"loan"**.
- In *"bank of the river"*, **"bank"** pays strong attention to **"river"**.

### 2. Transformer Architecture: Encoders & Decoders
Transformers consist of two main components:
- **Encoder**: Reads and understands context from text (e.g., BERT). Best for classification, NER, search.
- **Decoder**: Generates new text token by token (e.g., GPT). Best for story generation, code completion.
- **Encoder-Decoder**: Reads input text and generates output text (e.g., T5, BART). Best for translation, summarization.

### 3. Pre-training vs. Fine-tuning
- **Pre-training**: Training a massive Transformer model on billions of sentences (unlabeled text) to learn language structure.
- **Fine-tuning**: Adapting the pre-trained model to a specific task (e.g., classifying customer support tickets) using a small labeled dataset.
